```python
import os
import pandas as pd
```

### English

```python
data = []

with open('./data/dev-set-all-lang-with-gold-labels/EN/subtask-3-annotations.txt') as file:
    for line in file:
        line = line.split('\t')
        article_id = line[0]
        dominant_narrative = line[1]
        sub_narratives = line[2].strip("\n")
        ground_truth = line[3]

        data.append({'article_id': article_id, 'dominant_narrative': dominant_narrative, 'sub_narratives': sub_narratives, 'ground_truth': ground_truth})

df = pd.DataFrame(data)

for index, row in df.iterrows():
    path = os.path.join("data/dev-set-all-lang-with-gold-labels/EN/subtask-3-documents/", row['article_id'])

    with open(path) as file:
        text = file.read()
        df.at[index, 'text'] = text
```
