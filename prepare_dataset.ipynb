```python
import os
import pandas as pd

pd.set_option('display.max_colwidth', 1000)
```

```python
data = []

with open('SemEval 2025 Target Dec Release/EN/subtask-3-annotations.txt') as file:
    for line in file:
        line = line.split('\t')
        article_id = line[0]
        dominant_narrative = line[1]
        sub_narratives = line[2].strip("\n")
        ground_truth = line[3]

        data.append({'article_id': article_id, 'dominant_narrative': dominant_narrative, 'sub_narratives': sub_narratives, 'ground_truth': ground_truth})

df = pd.DataFrame(data)

for index, row in df.iterrows():
    path = os.path.join("SemEval 2025 Target Dec Release/EN/raw-documents", row['article_id'])

    with open(path) as file:
        text = file.read()
        df.at[index, 'text'] = text

df.to_csv('SemEval 2025 Test Data/final_datasets/eng_gold_label_data.csv', index=False)
```
